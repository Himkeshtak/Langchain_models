# Custom Tools

```python
from langchain_core.tools import tool
```

```python
# step -1 create the function
def multiply(a, b):

    """Multiply two numbers"""
    return a * b
```

```python
# step-2 give the hint
def multiply(a:int, b: int) -> int:
  """Multiply two numbers"""
  return a*b
```

```python
# step-3 add the decorator

@tool
def multiply(a:int, b: int) -> int:
  """Multiply two numbers"""
  return a*b
```

```python
result = multiply.invoke({'a': 3,'b':6})
```

```python
print(result)
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args_schema.schema())
```

# Method-2 (Using Structured Tool and Pydantic)

```python
from langchain_community.tools import StructuredTool
from pydantic import Field, BaseModel
```

```python
class MultiplyInput(BaseModel):
  a: int = Field(required = True, description='first number')
  b: int = Field(required = True,description='second number')
```

```python
def multiply_func(a: int,b: int)->int:
  return a*b
```

```python
multiply_tool = StructuredTool.from_function(
    func = multiply_func,
    name = 'multiply',
    description = 'Multiply two numbers',
    args_schema = MultiplyInput
)
```

```python
result = multiply_tool.invoke({'a': 3,'b':6})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)
print(multiply_tool.args_schema.schema)
```

# ToolKit

```python
from langchain_community.tools import tool

@tool
def multiply(a: float, b: float) -> float:
    """Multiply two numbers."""
    return a * b
@tool
def add(a: float, b: float)->float:
  """Add two numbers"""
  return a+b
```

```python
class MathToolkit:
  def get_tool(self):
    return [add, multiply]
```

```python
toolkit = MathToolkit()
tools = toolkit.get_tool()

for tool in tools:
  print(tool.name,"=>",tool.description)
```
