```python
import faiss
import numpy as np

d = 128  # 128 dimensioned vector embedding
nb = 100000 #it is the no. of random embedding vectors we wil generate
nq = 10

xb = np.random.random((nb, d)).astype('float32')
xq = np.random.random((nq, d)).astype('float32')
```

```python
index = faiss.IndexFlatL2(d)
```

```python
#This step loads all your database vectors into the index for searching.
#Each vector is stored in the order added.
#IDs are assigned as their position in the array.
index.add(xb)
```

```python
D, I = index.search(xq, k=5)
```

```python
print("Distances (D):")
display(D)
print("\nIndices (I):")
display(I)
```
